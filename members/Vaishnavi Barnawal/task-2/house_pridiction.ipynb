If we evaluate a model only on training data, it may appear to perform well because it has already learned from those examples. This does not tell us reliably how well it will predict prices for houses it has not seen before. Therefore, we keep separate testing data to evaluate the model on unseen examples.

# STEP - 06

```python
#Task1
x_train_mean=np.mean(x_train)
y_train_mean=np.mean(y_train)
print("mean of training sizes:- ",x_train_mean)
print("mean of training prices:- ",y_train_mean)
```

```python
#Task2
numerator = np.sum((x_train - x_train_mean) * (y_train - y_train_mean))
print("numerator:- ",numerator)
```

```python
#Task3
denominator = np.sum((x_train - x_train_mean) ** 2)
print("denominator:- ",denominator)
```

```python
#Task4
w=numerator/denominator
print("slope (w):- ",w)
```

```python
#Task5
b = y_train_mean - w * x_train_mean

print("Intercept (b):", b)
```

```python
#Task6
def predict_price(house_size):
    predicted_price = w * house_size + b
    return predicted_price

price = predict_price(1300)

print("Predicted price:", price, "lakh")
```

#STEP - 07

```python
#Task1
prediction= w * x_test + b
print(prediction)
```

```python
#Task2
error=y_test - prediction
print(error)
```

```python
#Task3
mae = np.mean(np.abs(error))
print("MAE:", mae)
```

```python
#Task4
mse = np.mean(np.abs(error) ** 2)
print("MSE:", mse)
```

TASK6

**MAE is easier to interpret than MSE** because MAE is measured in the same unit as the actual house prices, which is lakh rupees. It tells us the average difference between the actual and predicted prices. MSE uses squared errors, so its unit is lakh rupees squared, which is less intuitive to understand.

In our dataset, both MAE and MSE are 0 because the model's predictions match all the test prices exactly. However, this small and simple dataset does not guarantee that the model will perform perfectly on real-world house prices.
