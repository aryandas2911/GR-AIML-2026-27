# House Price Prediction

```python
import numpy as np 
house_sizes = np.array([500,800,1000,1200,1500,
                        1800 , 2000, 2200,2500 , 2800 ])
house_prices = np.array([15, 24, 30 , 36 , 45 , 
                         54 , 60 , 66 , 75 , 84 ])
```

## Spliting data into training and testing sets

```python
x_train = house_sizes[:8]
y_train = house_prices[:8]

x_test = house_sizes[8:]
y_test = house_prices[8:]

print("x_train:", x_train)
print("y_train:", y_train)
print("x_test:", x_test)
print("y_test:", y_test)

print("Number of training examples:", x_train.size)
print("Number of test examples:", x_test.size)

print("Training Pairs (Size : Price):")
for size, price in zip(x_train, y_train):
    print(f"{size} sq ft : ${price}k")

print("\nTest Pairs (Size : Price):")
for size, price in zip(x_test, y_test):
    print(f"{size} sq ft : ${price}k")
```

```python
# 1. Calculate the mean of x_train and y_train
mean_x = np.mean(x_train)
mean_y = np.mean(y_train)

# 2. Calculate the numerator of the slope formula: sum((x - mean_x) * (y - mean_y))
numerator = np.sum((x_train - mean_x) * (y_train - mean_y))

# 3. Calculate the denominator of the slope formula: sum((x - mean_x)^2)
denominator = np.sum((x_train - mean_x) ** 2)

# 4. Calculate w (the slope)
w = numerator / denominator

# 5. Calculate b (the y-intercept) using b = mean_y - w * mean_x
b = mean_y - (w * mean_x)

print(f"Calculated slope (w): {w}")
print(f"Calculated intercept (b): {b}")

# 6. Write a Python function that accepts a house size and returns a predicted price
def predict_price(size):
    return w * size + b

# 7. Use your function to predict the price of a 1,300-square-foot house
house_size_to_predict = 1300
predicted_price = predict_price(house_size_to_predict)

print(f"Predicted Price for a {house_size_to_predict} sq ft house: ${predicted_price:.2f}k")
```

```python
# 1. Generate predictions for all test houses using our predict_price function
test_predictions = predict_price(x_test)
print("Predictions for test set:", test_predictions)

# 2. Subtract the predictions from the actual prices to calculate errors
# Error = Actual (y_test) - Predicted
errors = y_test - test_predictions
print("Errors:", errors)

# 3. Calculate Mean Absolute Error (MAE) using NumPy
# Take the absolute value of each error, then find the mean
mae = np.mean(np.abs(errors))
print(f"Mean Absolute Error (MAE): {mae:.2f}k")

# 4. Calculate Mean Squared Error (MSE) using NumPy
# Square each error, then find the mean
mse = np.mean(errors ** 2)
print(f"Mean Squared Error (MSE): {mse:.2f}")
```
