```python
import numpy as np
house_sizes = np.array([500, 800, 1000, 1200, 1500, 1800, 2000, 2200, 2500, 2800])
house_prices = np.array([15, 24, 30, 36, 45, 54, 60, 66, 75, 84])
print(house_sizes)
print(house_prices)

print("number of houses: ",house_sizes.size)

print("size of house_sizes:", house_sizes.size)
print("shape of house_sizes:", house_sizes.shape)

print("size of house_prices:", house_prices.size)
print("shape of house_prices:", house_prices.shape)
```

```python
#house size stats
print("mean:", np.mean(house_sizes))
print("minimum:", np.min(house_sizes))
print("maximum:", np.max(house_sizes))
print("median:", np.median(house_sizes))
print("standard deviation:", np.std(house_sizes))

#house price stats
print("mean:", np.mean(house_prices))
print("minimum:", np.min(house_prices))
print("maximum:", np.max(house_prices))
print("median:", np.median(house_prices))
print("standard deviation:", np.std(house_prices))

```

```python
size1 = house_sizes[0]
size2 = house_sizes[1]

price1 = house_prices[0]
price2 = house_prices[1]

size_diff = size2 - size1
price_diff = price2 - price1

print("size difference:", size_diff)
print("price difference:", price_diff)

w = price_diff / size_diff
print("slope :", w)

b = price1 - w * size1
print("bias :", b)

new_size = 1800
predicted_price = w * new_size + b

print("predicted price:", predicted_price)
```

```python
X_train = house_sizes[:8]
y_train = house_prices[:8]
X_test = house_sizes[8:]
y_test = house_prices[8:]

print("X_train:", X_train)
print("y_train:", y_train)
print("X_test:", X_test)
print("y_test:", y_test)

print("number of training examples:", X_train.size)
print("number of testing examples:", X_test.size)

print("training pairs (size, price):")
print(np.column_stack((X_train, y_train)))

print("testing pairs (size, price):")
print(np.column_stack((X_test, y_test)))
```

```python
x_mean = np.mean(X_train)
y_mean = np.mean(y_train)

print("mean of house sizes:", x_mean)
print("mean of house prices:", y_mean)

num = np.sum((X_train - x_mean) * (y_train - y_mean))
print("numerator:", num)

denom = np.sum((X_train - x_mean) ** 2)
print("denominator:", denom)

w = num / denom
print("slope:", w)

b = y_mean - w * x_mean
print("bias:", b)

def predict_price(size):
    return w * size + b

predicted_price = predict_price(1300)
print("predicted price:", predicted_price)
```

```python
y_pred = predict_price(X_test)

print("actual prices:", y_test)
print("predicted prices:", y_pred)

errors = y_test - y_pred

print("prediction errors:", errors)

mae = np.mean(np.abs(errors))
print("mean absolute error:", mae, "lakhs")

mse = np.mean(errors ** 2)
print("mean squared error:", mse, "lakhs squared")
```

```python
size = float(input("enter house size in square feet: "))
if size <= 0:
    print("enter a positive house size.")
else:
    predicted_price = predict_price(size)
    print("predicted house price:", predicted_price, "lakhs")
```
