```python
import numpy as np
house_sizes = np.array(
    [    500, 800, 1000, 1200, 1500,  
      1800, 2000, 2200, 2500, 2800])
house_prices = np.array(
    [    15, 24, 30, 36, 45,   
      54, 60, 66, 75, 84 ])      
print(house_sizes) 
print(house_prices) 

house_sizes[2]
house_prices[2]
print(house_prices.size)
print(house_prices.shape)
```

```python
house_sizes=np.append(house_sizes,3000)
house_prices=np.append(house_prices,90)
print(house_sizes)
print(house_prices)
```

```python
house_size_avg=np.mean(house_sizes)
house_size_min=np.min(house_sizes)
house_size_max=np.max(house_sizes)
house_size_std=np.std(house_sizes)
house_size_med=np.median(house_sizes)
print(house_size_avg)
print(house_size_min)
print(house_size_max)
print(house_size_std)
print(house_size_med)
```

```python
house_prices_avg=np.mean(house_prices)
house_prices_min=np.min(house_prices)
house_prices_max=np.max(house_prices)
house_prices_std=np.std(house_prices)
house_prices_med=np.median(house_prices)
print(house_prices_avg)
print(house_prices_min)
print(house_prices_max)
print(house_prices_std)
print(house_prices_med)
```

```python
size_diff=house_sizes[2]-house_sizes[0]
print(size_diff)
price_diff=house_prices[2]-house_prices[0]
print(price_diff)
```

```python
w=price_diff/size_diff
print(w)
```

```python
b=house_prices[0]-w*house_sizes[0]
print(b)
```

```python
def predict_price(size):
    predict_price=w*size+b
    return(predict_price)

print(predict_price(1300))
```

```python
X_train = house_sizes[:8] 
y_train = house_prices[:8] 
X_test = house_sizes[9:] 
y_test = house_prices[9:]
```

```python
print(X_train)
print(y_train)
print(X_test)
print(y_test)
```

```python
print(X_train.size) 
print(y_train.size) 
print(X_test.size)
print(y_test.size)
```

```python
x_mean=np.mean(X_train) 
y_mean=np.mean(y_train)
num=((X_train-x_mean)*(y_train-y_mean)).sum()
den=((X_train - x_mean)**2).sum()
w=num/den
print(w)
b=y_mean-w*x_mean
print(b)
```

```python
y_pred=[predict_price(size) for size in X_test]
print(y_pred)
```

```python
error = y_test - y_pred
print(error)

mae = np.abs(error)
MAE = np.mean(mae)
print(MAE)

MSE = np.mean(error**2)
print(MSE)
```
