```python
import numpy as np
print(np.__version__)
```

```python
# Step 2: Create your dataset
house_sizes = np.array([500, 800, 1000, 1200, 1500, 1800, 2200, 2200, 2500, 2800])
house_prices = np.array([15, 24, 30, 36, 45, 54, 66, 66, 75, 84])

# Print both arrays
print("House Sizes (sq ft):", house_sizes)
print("House Prices (lakhs):", house_prices)

# Print size and shape
print("Array Size:", house_sizes.size)
print("Array Shape:", house_sizes.shape)

# Accessing the third house (index 2)
print(f"Third house -> Size: {house_sizes[2]} sq ft, Price: {house_prices[2]} lakhs")

# Add one more house to the dataset
house_sizes = np.append(house_sizes, 3000)
house_prices = np.append(house_prices, 90)
print("Updated House Sizes after adding a new house:", house_sizes)
```

```python
# Step 3: Explore the data (EDA)
print("=== House Sizes Statistics ===")
print("Average (Mean) Size:", np.mean(house_sizes))
print("Minimum Size:", np.min(house_sizes))
print("Maximum Size:", np.max(house_sizes))
print("Median Size:", np.median(house_sizes))
print("Standard Deviation of Sizes:", round(np.std(house_sizes), 2))

print("\n=== House Prices Statistics ===")
print("Average (Mean) Price:", np.mean(house_prices))
print("Minimum Price:", np.min(house_prices))
print("Maximum Price:", np.max(house_prices))
print("Median Price:", np.median(house_prices))
print("Standard Deviation of Prices:", round(np.std(house_prices), 2))
```

```python
# Step 4: Investigate the relationship
# Let's check the difference between the first two houses
size_diff = house_sizes[1] - house_sizes[0]
price_diff = house_prices[1] - house_prices[0]

print(f"Size difference between house 2 and house 1: {size_diff} sq ft")
print(f"Price difference between house 2 and house 1: {price_diff} lakhs")

# Relationship check: how much price changes per 100 sq ft
price_per_100sqft = (price_diff / size_diff) * 100
print(f"Every additional 100 square feet corresponds to an additional {price_per_100sqft:.1f} lakhs.")
```

```python
# Step 5: Split the data into training and testing sets
# Using the original 10-house dataset arrays
orig_sizes = np.array([500, 800, 1000, 1200, 1500, 1800, 2200, 2200, 2500, 2800])
orig_prices = np.array([15, 24, 30, 36, 45, 54, 66, 66, 75, 84])

X_train = orig_sizes[:8]
y_train = orig_prices[:8]
X_test = orig_sizes[8:]
y_test = orig_prices[8:]

print("X_train (Training sizes):", X_train)
print("y_train (Training prices):", y_train)
print("X_test (Test sizes):", X_test)
print("y_test (Test prices):", y_test)

print(f"\nNumber of training examples: {len(X_train)}")
print(f"Number of test examples: {len(X_test)}")
```

```python
# Step 6: Build your first prediction model (Linear Regression)
x_mean = np.mean(X_train)
y_mean = np.mean(y_train)

# Calculate slope (w) and intercept (b) using the closed-form formulas
numerator = np.sum((X_train - x_mean) * (y_train - y_mean))
denominator = np.sum((X_train - x_mean) ** 2)

w = numerator / denominator
b = y_mean - (w * x_mean)

print(f"Calculated Slope (w): {w:.4f}")
print(f"Calculated Intercept (b): {b:.4f}")

# Define the prediction function
def predict(size):
    return w * size + b

# Predict the price of a 1,300-square-foot house
sample_size = 1300
predicted_price = predict(sample_size)
print(f"Predicted price for a {sample_size}-sq-ft house: {predicted_price:.2f} lakhs")
```

```python
# Step 7: Evaluate your model
y_pred = predict(X_test)

print("Actual test prices:", y_test)
print("Predicted test prices:", np.round(y_pred, 2))

# Calculate MAE and MSE using NumPy
mae = np.mean(np.abs(y_pred - y_test))
mse = np.mean((y_pred - y_test) ** 2)

print(f"\nMean Absolute Error (MAE): {mae:.4f} lakhs")
print(f"Mean Squared Error (MSE): {mse:.4f}")
```

```python
# Step 8: Make the program interactive
print("=== Interactive House Price Predictor ===")
user_input = input("Enter house size in square feet (e.g., 1600): ")

try:
    house_size_val = float(user_input)
    predicted_price = predict(house_size_val)
    print(f"Predicted house price: {predicted_price:.2f} lakhs")
except ValueError:
    print("Invalid input! Please enter a valid number instead of text.")
```
