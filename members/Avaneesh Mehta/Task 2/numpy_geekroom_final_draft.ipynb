```python
import numpy as np

# Tasks of Step 2
house_sizes = np.array([500,800,1000,1200,1500,1800,2000,2200,2500,2800])
house_prices = np.array([15,24,30,36,45,54,60,66,75,84])
print(house_sizes) #to print array house_sizes
print(house_prices) #to print array house_prices
x = house_sizes.size
print("Number of houses in the dataset",x)
y = house_prices.size
print(y) #to print size of array house_prices
print(house_sizes.shape) #to print shape of array house_sizes
print(house_prices.shape) #to print shape of array house_prices
print("Size of third house is",house_sizes[2:3],"square feet")
print("Price of third house is",house_prices[2:3],"lakhs")

#following code is used to ask user the size and price of a house and then add
#those into the respective arrays.
user_size = int(input("Enter house size (in sq ft): "))
user_price = int(input("Enter house price (in lakhs): "))
house_sizes = np.append(house_sizes, user_size)
house_prices = np.append(house_prices, user_price)
print("Updated house_sizes:", house_sizes)
print("Updated house_prices:", house_prices)

# Tasks of Step 3
print("Average house size is",house_sizes.mean(),"sq ft")
print("Maximum size of house is",house_sizes.max(),"sq ft")
print("Minimum size of house is",house_sizes.min(),"sq ft")
print("Median house size is",np.median(house_sizes),"sq ft")
print("Standard deviation of sizes of houses is",house_sizes.std(),"sq ft")
print("Average house price is",house_prices.mean(),"lakhs")
print("Maximum price of house is",house_prices.max(),"lakhs")
print("Minimum price of house is",house_prices.min(),"lakhs")
print("Median house price is",np.median(house_prices),"lakhs")
print("Standard deviation of cost of houses is",house_prices.std(),"lakhs")

#Average of price of houses tells us the value of one house if the total value
#of all houses were distributed equally
#the median is the exact middle value of dataset when it is arranged in
#ascending or descending order
#If we add an extremely expensive house in the dataset, value of average price
#would go very up, as mean depends on sum of all prices which increases due to
#final price
#whereas median depends on middle value of ordered dataset, hence it would
#usually change less

# Tasks of Step 4
house1_size = house_sizes[0]
house2_size = house_sizes[1]
house1_price = house_prices[0]
house2_price = house_prices[1]
print("Difference between sizes of first two houses is= ", house2_size-house1_size)
print("Difference between prices of first two houses is= ", house2_price-house1_price)
w = (house2_price-house1_price)/(house2_size-house1_size)
#price difference is directly proportional to size difference or ratio of price
#difference and size difference is constant which is 0.03 for this dataset
print(w)
#following is code for python expression that predicts a price using the relation
b = house1_price - (w * house1_size)
print(f"b (intercept) = {b}")
size = 1300
predicted_price = w*size + b
print("predicted price is=",predicted_price)

# Tasks of Step 5
X_train = house_sizes[:8]
Y_train = house_prices[:8]
X_test = house_sizes[8:]
Y_test = house_prices[8:]

#To print all four arrays
print(X_train)
print(Y_train)
print(X_test)
print(Y_test)

#To count training and testing examples
print(X_train.size)
print(Y_train.size)
print(X_test.size)
print(Y_test.size)

#To check if each input is still paired with correct target price
for i in range(len(house_sizes)):
    print(f"House {i+1}: Size = {house_sizes[i]} sq ft -> Price = {house_prices[i]} Lakh")

#testing on training data can give misleading impression because the model would
#give a better result on the data which it was trained on, hence if a new set of
#data would be given by the user, the model might show a high rate of inaccuracy
#while solving the problem with a different pattern from the training data

# Tasks of Step 6
x_mean = np.mean(X_train)
y_mean = np.mean(Y_train)
print("Mean of X_train is",x_mean)
print("Mean of y_train is",y_mean)

numerator = np.sum((X_train-x_mean)*(Y_train-y_mean))
denominator = np.sum((X_train-x_mean)**2)
print("Numerator of slope formula is",numerator)
print("Denominator of slope formula is",denominator)

w = numerator/denominator
b = y_mean - w*x_mean
print("w (slope) =",w)
print("b (intercept) =",b)

def predict_price(size):
    return w*size + b

print("Predicted price of a 1300 sq ft house is",predict_price(1300),"lakhs")

# Tasks of Step 7
predicted_prices = w*X_test + b
errors = Y_test - predicted_prices
print("Actual prices are",Y_test)
print("Predicted prices are",predicted_prices)
print("Errors are",errors)

MAE = np.mean(np.abs(errors))
MSE = np.mean(errors**2)
print("Mean Absolute Error (MAE) is",MAE,"lakhs")
print("Mean Squared Error (MSE) is",MSE,"lakhs squared")

# MAE is easier to interpret because it is in the original unit of house prices.
# MSE gives larger errors more importance because the errors are squared.

# Tasks of Step 8
#To ask the user for a house size and predict its price
try:
    new_size = float(input("Enter house size in square feet: "))
    if new_size <= 0:
        print("Invalid input. House size must be greater than zero.")
    else:
        print(f"Predicted house price: {predict_price(new_size):.2f} lakhs")
except ValueError:
    print("Invalid input. Please enter a numeric house size.")

# Tasks of Step 9

# Challenge A: Find unusual houses
#To check the mean and median before adding an unusually expensive house
mean_before = np.mean(house_prices)
median_before = np.median(house_prices)
print("Mean price before adding unusual house is",mean_before,"lakhs")
print("Median price before adding unusual house is",median_before,"lakhs")

#To add a house with an unusually high price
#For testing, enter 3000 as size and 250 as price
newhouse_size = int(input("Enter the size of an unusual house: "))
newhouse_price = int(input("Enter the unusually high price of the house: "))
house_sizes = np.append(house_sizes, newhouse_size)
house_prices = np.append(house_prices, newhouse_price)
print("Added an unusual house of size",newhouse_size,"sq ft and price",newhouse_price,"lakhs")

#To check the mean and median after adding an unusually expensive house
mean_after = np.mean(house_prices)
median_after = np.median(house_prices)
print("Mean price after adding unusual house is",mean_after,"lakhs")
print("Median price after adding unusual house is",median_after,"lakhs")

#The mean usually increases a lot because it uses every price in the sum.
#The median depends on the middle value, so it is less affected by an extreme price.
#An unusual house price can also change the slope and intercept of the model.
#Linear regression gives more importance to large errors because it squares them.

# Challenge B: Shuffle the dataset
#One shared permutation keeps each house size paired with its correct price
np.random.seed(42)
indices = np.random.permutation(len(house_sizes))
house_sizes = house_sizes[indices]
house_prices = house_prices[indices]
print("Shuffled house sizes:",house_sizes)
print("Shuffled house prices:",house_prices)

#To split the shuffled data into approximately 60 percent training data and 40 percent testing data
split_index = int(0.6*len(house_sizes))
X_train = house_sizes[:split_index]
Y_train = house_prices[:split_index]
X_test = house_sizes[split_index:]
Y_test = house_prices[split_index:]
print("New X_train:",X_train)
print("New Y_train:",Y_train)
print("New X_test:",X_test)
print("New Y_test:",Y_test)

#To calculate w and b again using the shuffled training data
x_mean = np.mean(X_train)
y_mean = np.mean(Y_train)
numerator = np.sum((X_train-x_mean)*(Y_train-y_mean))
denominator = np.sum((X_train-x_mean)**2)
w = numerator/denominator
b = y_mean - w*x_mean
print("New w (slope) =",w)
print("New b (intercept) =",b)

#To update the prediction function with the new values of w and b
def predict_price(size):
    return w*size + b

#To evaluate the model on the shuffled test data
predicted_prices = w*X_test + b
errors = Y_test - predicted_prices
print("Actual test prices are",Y_test)
print("Predicted test prices are",predicted_prices)
print("Errors are",errors)
MAE = np.mean(np.abs(errors))
MSE = np.mean(errors**2)
print("New Mean Absolute Error (MAE) is",MAE,"lakhs")
print("New Mean Squared Error (MSE) is",MSE,"lakhs squared")

# Tasks of Step 10

#Objective:
#The objective of this project is to predict house prices using house sizes.

#Dataset:
#House size is the feature and it is measured in square feet.
#House price is the target and it is measured in lakhs of rupees.

#Method:
#I used NumPy to calculate the mean of training sizes and prices.
#I calculated the slope using the numerator and denominator of the slope formula.
#I calculated the intercept using the means, slope and training data.
#The prediction equation is predicted price = w * house size + b.

#Results:
#The program predicts prices for the test data and calculates their errors.
#It also calculates MAE and MSE to evaluate the predictions.
#The results before and after shuffling can be compared using the printed output.

#Challenges completed:
#Challenge A: Added an unusually expensive house and compared the mean and median.
#Challenge B: Shuffled the data while keeping each size paired with its price.
#After shuffling, I split the data again and retrained and evaluated the model.

#Reflection:
#I learned how to use NumPy for data analysis and implement linear regression.
#I also learned how to make predictions and calculate MAE and MSE.

#Limitations:
#1. The dataset is small and may not represent real-world house prices.
#2. The model uses only house size and ignores other factors such as location,
#age, condition and facilities.
```
