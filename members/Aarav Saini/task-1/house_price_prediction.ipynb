## Step 1 - setup

```python
import numpy as np

print("numpy version:", np.__version__)
```

## Step 2 - dataset

```python
house_sizes = np.array([500, 800, 1000, 1200, 1500,1800, 2000, 2200, 2500, 2800])
house_prices = np.array([15, 24, 30, 36, 45,54, 60, 66, 75, 84])

print(house_sizes)
print(house_prices)

print("number of houses:", len(house_sizes))
print("size:", house_sizes.size, house_prices.size)
print("shape:", house_sizes.shape, house_prices.shape)
print("third house ->", house_sizes[2], "sqft,", house_prices[2], "lakh")

new_sizes = np.append(house_sizes, 3000)
new_prices = np.append(house_prices, 90)
print("after adding a house:", len(new_sizes), len(new_prices))
print(new_sizes)
print(new_prices)
```

## Step 3 - explore the data

```python
size_mean = np.mean(house_sizes)
size_min = np.min(house_sizes)
size_max = np.max(house_sizes)
size_median = np.median(house_sizes)
size_std = np.std(house_sizes)

price_mean = np.mean(house_prices)
price_min = np.min(house_prices)
price_max = np.max(house_prices)
price_median = np.median(house_prices)
price_std = np.std(house_prices)

print("sizes  -> mean", size_mean, "min", size_min, "max", size_max, "median", size_median, "std", round(size_std, 2))
print("prices -> mean", price_mean, "min", price_min, "max", price_max, "median", price_median, "std", round(price_std, 2))

print("average size is", size_mean, "sqft")
print("smallest is", size_min, "and biggest is", size_max, "sqft")
print("median size", size_median, "sqft")
print("std of sizes is", round(size_std, 2), "which shows how far houses are from the average")
print("average price is", price_mean, "lakh")
print("median price is", price_median, "lakh")
```

## Step 4 - relationship

```python
diff_price = house_prices[5] - house_prices[1]
diff_size = house_sizes[5] - house_sizes[1]
print("price difference:", diff_price)
print("size difference:", diff_size)
print("price per sqft:", diff_price / diff_size)
print("so every 100 sqft adds", diff_price / diff_size * 100, "lakh")

def quick_predict(size):
    return size * 0.03

print("quick guess for 1000 sqft:", quick_predict(1000))
```

## Step 5 - train test split

```python
X_train = house_sizes[:8]
y_train = house_prices[:8]
X_test = house_sizes[8:]
y_test = house_prices[8:]

print(X_train)
print(y_train)
print(X_test)
print(y_test)
print("train count:", len(X_train), "test count:", len(X_test))
for i in range(len(X_train)):
    print(X_train[i], "->", y_train[i])
for i in range(len(X_test)):
    print(X_test[i], "->", y_test[i])
```

## Step 6 - linear regression

```python
x_mean = np.mean(X_train)
y_mean = np.mean(y_train)
print("x mean:", x_mean, "y mean:", y_mean)

numerator = np.sum((X_train - x_mean) * (y_train - y_mean))
denominator = np.sum((X_train - x_mean) ** 2)
print("numerator:", numerator)
print("denominator:", denominator)

w = numerator / denominator
b = y_mean - w * x_mean
print("w =", w)
print("b =", b)

def predict_price(size):
    return w * size + b

print("price of 1300 sqft house:", predict_price(1300))
```

## Step 7 - evaluate

```python
predictions = predict_price(X_test)
print("predicted:", predictions)
print("actual:", y_test)

errors = y_test - predictions
print("errors:", errors)

mae = np.mean(np.abs(errors))
mse = np.mean(errors ** 2)
print("MAE:", mae)
print("MSE:", mse)
```

## Step 8 - interactive

```python
while True:
    text = input("Enter house size in square feet: ")
    try:
        size = float(text)
        if size <= 0:
            print("size has to be more than 0")
            continue
        print("Predicted house price:", round(predict_price(size), 2), "lakh")
        break
    except ValueError:
        print("thats not a number, try again")
```

## Challenge A - unusual houses

```python
big_sizes = np.append(house_sizes, 2600)
big_prices = np.append(house_prices, 400)
print("before -> mean", np.mean(house_prices), "median", np.median(house_prices))
print("after  -> mean", np.mean(big_prices), "median", np.median(big_prices))
print("mean jumped a lot but median moved much less. outliers can drag the line a lot when fitting so the model gets worse for normal houses")

big_x_mean = np.mean(big_sizes)
big_y_mean = np.mean(big_prices)
big_w = np.sum((big_sizes - big_x_mean) * (big_prices - big_y_mean)) / np.sum((big_sizes - big_x_mean) ** 2)
big_b = big_y_mean - big_w * big_x_mean
print("slope with outlier:", big_w, "intercept:", big_b)
print("slope without outlier:", w, "intercept:", b)
```

## Challenge B - shuffle

```python
np.random.seed(1)
perm = np.random.permutation(len(house_sizes))
shuffled_sizes = house_sizes[perm]
shuffled_prices = house_prices[perm]
print(shuffled_sizes)
print(shuffled_prices)

s_train_x = shuffled_sizes[:8]
s_train_y = shuffled_prices[:8]
s_test_x = shuffled_sizes[8:]
s_test_y = shuffled_prices[8:]

sx_mean = np.mean(s_train_x)
sy_mean = np.mean(s_train_y)
sw = np.sum((s_train_x - sx_mean) * (s_train_y - sy_mean)) / np.sum((s_train_x - sx_mean) ** 2)
sb = sy_mean - sw * sx_mean
s_pred = sw * s_test_x + sb
print("test sizes:", s_test_x, "actual:", s_test_y, "predicted:", s_pred)
print("MAE after shuffle:", np.mean(np.abs(s_test_y - s_pred)))
```
